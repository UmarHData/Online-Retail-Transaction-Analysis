# Extract, Transform, Load - ETL

Import Python libraries 

```python
import pandas as pd
import numpy as np
```

Import raw dataset

```python
# Load the raw dataset into a pandas DataFrame
df = pd.read_csv('Data/Online Retail.csv')

# Display the first five rows to check that the dataset loaded correctly
df.head()
```

### Preview the Dataset
The first five and last five rows of the dataset are displayed to get an understanding of the data, including column names, values and overall structure of the dataset.

---

```python
df.head()
```

```python
df.tail()
```

The first and last five rows show that each row represents an individual transaction. This initial inspection helps confirm that the dataset has loaded correctly and provides an overview of the types of information available for further investigation and cleaning.

### Dataset Information
The dataset information is reviewed to understand its structure, column data types and whether any missing values are present before further investigation and cleaning.

```python
# Display the dataset structure, column data types and non-null counts
df.info()
```

The dataset contains 541,909 rows and 8 columns. The Description column contains 1,454 missing values, while the remaining columns contain no missing values. InvoiceDate is currently stored as a string and will need to be converted to datetime for time-based analysis.

### Missing Values
I will check the dataset for missing values in each column to identify any incomplete records that may require cleaning.

```python
# Count the number of missing values in each column
df.isnull().sum()
```

The missing-value check identified 1,454 missing values in the Description column. No missing values were identified in the other columns.

### Investigating Missing Product Descriptions

The Description column contains 1,454 missing values. I will examine these values to determine whether they contain useful transaction information.

```python
# Display transactions where the product description is missing
missing_description = df[df['Description'].isna()]

missing_description.head(20)
```

After the initial investigation the dataset shows that rows with missing product descriptions have a UnitPrice of 0.0 and include both positive and negative quantities.

```python
# Count missing-description transactions with a UnitPrice of 0.0
(missing_description['UnitPrice'] == 0.0).sum()
```

This shows all 1,454 rows with a missing product description also have a UnitPrice of 0.0. I will investigate quantity and customer patterns before deciding whether to remove them.

```python
# Count negative quantities among rows with missing descriptions
(missing_description['Quantity'] < 0).sum()
```

```python
# Count positive quantities among rows with missing descriptions
(missing_description['Quantity'] > 0).sum()
```

```python
# Check how many unique customers appear in rows with missing descriptions
missing_description['CustomerID'].nunique()
```

I found that all 1,454 rows with missing product descriptions have a UnitPrice of 0. Of these rows, 862 have negative quantities and 592 have positive quantities. All rows are also associated with a single CustomerID. This pattern shows that these may not represent normal revenue-generating customer transactions.

```python
# Display the most common invoice numbers among rows with missing descriptions
missing_description['InvoiceNo'].value_counts().head(20)
```

```python
# Count the number of unique invoices among rows with missing descriptions
missing_description['InvoiceNo'].nunique()
```

```python
# Display the most common StockCodes among rows with missing descriptions
missing_description['StockCode'].value_counts().head(20)
```

```python
# Investigate one affected StockCode across the full dataset
df[df['StockCode'] == '35965'][
    ['InvoiceNo', 'StockCode', 'Description', 'Quantity', 'UnitPrice']
].head(20)
```

I investigated further and found that each row has a unique InvoiceNo and individual StockCodes showed that the same products appear elsewhere in the dataset with valid descriptions and positive selling prices. 

For example, StockCode 35965 appears as "FOLKART HEART NAPKIN RINGS" with positive UnitPrice values, while its missing-description row has a UnitPrice of 0 and a negative quantity.

This pattern shows that the missing-description rows are more likely to represent non-revenue activity rather than normal customer sales.

### Removing missing product descriptions

```python
# Remove rows with missing descriptions identified as zero-value, non-standard transactions
df.dropna(subset=['Description'], inplace=True)
```

```python
# Check the number of missing values in each column after cleaning
df.isnull().sum()
```

```python
# Check the number of rows and columns after removing missing product descriptions
df.shape
```

I have 1,454 rows with missing product descriptions, the dataset now contains 540,455 rows and 8 columns.

```python
# Reset the index after removing rows
df.reset_index(drop=True, inplace=True)
```

### Check for duplicated rows
I will check for duplicate rows in the dataset. Duplicate transactions could lead to double counting and may distort sales and revenue.

```python
# Count the number of duplicated rows in the dataset
df.duplicated().sum()
```

```python
# Calculate and round the percentage of duplicated rows
round((df.duplicated().sum() / len(df)) * 100, 2)
```

### Investigating duplicate rows
I checked for dupliclate rows and found 5,268 duplicated rows in the dataset, representing approximately 0.97% of the dataset. These rows will be inspected before removal to determine whether they represent duplicate transactions or legitimate repeated purchases.

```python
# Display duplicated rows for investigation
df[df.duplicated(keep=False)].head(20)
```

Inspection showed that these rows are identical across all columns, including InvoiceNo, StockCode, Quantity, InvoiceDate, UnitPrice and CustomerID.

### Remove duplicated rows

```python
# Remove duplicate rows while keeping the first occurrence
df.drop_duplicates(inplace=True)
```

```python
# Confirm no duplicate rows remain in the dataset
df.duplicated().sum()
```

```python
# Check the dataset dimensions after removing duplicates
df.shape
```

```python
# Reset the index after removing duplicate rows
df.reset_index(drop=True, inplace=True)
```

After removing 5,268 duplicate rows, the dataset contains 535,187 rows and 8 columns.
